```python
# ==============================================================================
# STEP 0: DOWNLOAD CIFAR-100 IF NEEDED
# ==============================================================================
import os, urllib.request, tarfile

CIFAR_DIR = "/content/cifar-100-python"

if not os.path.isdir(CIFAR_DIR):
    print("Downloading CIFAR-100 (~170 MB)...")
    urllib.request.urlretrieve(
        "https://www.cs.toronto.edu/~kriz/cifar-100-python.tar.gz",
        "/content/cifar-100-python.tar.gz"
    )
    print("Extracting...")
    with tarfile.open("/content/cifar-100-python.tar.gz", "r:gz") as tar:
        tar.extractall("/content/")
    os.remove("/content/cifar-100-python.tar.gz")
    print("Done.")
else:
    print("CIFAR-100 already present.")

for f in ["train", "test", "meta"]:
    p = os.path.join(CIFAR_DIR, f)
    print(f"  {f}: {'✅ OK' if os.path.isfile(p) else '❌ MISSING'}")

assert os.path.isdir(CIFAR_DIR), "CIFAR-100 download failed"
print("✅ CIFAR-100 ready\n")

# ==============================================================================
# CGL-AARM v5.1 — CONFIDENCE-CALIBRATED AEFR — FINAL FULL RUN (COLAB)
# ==============================================================================

import time, random, pickle, warnings, json, contextlib
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torchvision
from torch.utils.data import DataLoader, TensorDataset
from scipy import stats

warnings.filterwarnings("ignore")

# ==============================================================================
# CONFIG
# ==============================================================================
SMOKE_TEST = False

if SMOKE_TEST:
    SEEDS, MEMORY_LEVELS, RUN_ABLATION = [42, 100], [0], False
else:
    SEEDS         = [42, 100, 2024, 777, 999]
    MEMORY_LEVELS = [0, 50, 100, 200]
    RUN_ABLATION  = True

NUM_TASKS        = 10
CLASSES_PER_TASK = 10
FEATURE_DIM      = 512

EXPERT_EPOCHS    = 12
ROUTER_EPOCHS    = 12

VAL_FRAC         = 0.15
TRAIN_BATCH      = 1024
EVAL_BATCH       = 4096

MAHA_PCA_DIM     = 64

TOP1_CONFIDENCE  = 0.65
TOP2_CONFIDENCE  = 0.35
CALIB_POWER      = 2.0

TASK_ORDERS = {
    "Order_A": list(range(10)),
    "Order_B": list(range(9, -1, -1))
}

LIVING_TASKS   = [0, 1, 4, 6, 7]
ARTIFACT_TASKS = [2, 3, 5, 8, 9]

OUTPUT_DIR = "/content/CGL_AARM_v51"
os.makedirs(OUTPUT_DIR, exist_ok=True)


def log(m=""):
    print(f"[{time.strftime('%H:%M:%S')}] {m}", flush=True)


# ==============================================================================
# DEVICE
# ==============================================================================
device  = torch.device("cuda" if torch.cuda.is_available() else "cpu")
USE_AMP = device.type == "cuda"

if USE_AMP:
    torch.backends.cudnn.benchmark        = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32       = True
    log(f"✅ GPU: {torch.cuda.get_device_name(0)} "
        f"({torch.cuda.get_device_properties(0).total_memory/1e9:.1f} GB)")
else:
    log("⚠️  NO GPU — go to Runtime → Change runtime type → T4 GPU")

log(f"Device={device} | SMOKE_TEST={SMOKE_TEST} | PyTorch={torch.__version__}")


# ==============================================================================
# RANDOM SEED
# ==============================================================================
def set_seed(s):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)
    if USE_AMP:
        torch.cuda.manual_seed_all(s)


# ==============================================================================
# AMP
# ==============================================================================
def make_scaler():
    if not USE_AMP:
        class D:
            def scale(self, l): return l
            def step(self, o): o.step()
            def update(self): pass
        return D()
    try:
        return torch.amp.GradScaler("cuda")
    except:
        return torch.cuda.amp.GradScaler()


def amp_ctx():
    if not USE_AMP:
        return contextlib.nullcontext()
    try:
        return torch.autocast("cuda", dtype=torch.float16)
    except:
        return torch.cuda.amp.autocast()


# ==============================================================================
# DATA + FEATURES
# ==============================================================================
def unpickle(fp):
    with open(fp, "rb") as f:
        return pickle.load(f, encoding="bytes")


def load_cifar100():
    def _l(n):
        d = unpickle(os.path.join(CIFAR_DIR, n))
        return (
            d[b"data"].reshape(-1, 3, 32, 32).astype(np.float32) / 255.0,
            np.array(d[b"fine_labels"], dtype=np.int64)
        )
    Xtr, Ytr = _l("train")
    Xte, Yte = _l("test")
    mn = np.array([.5071, .4867, .4408], dtype=np.float32)[:, None, None]
    st = np.array([.2675, .2565, .2761], dtype=np.float32)[:, None, None]
    return (Xtr - mn) / st, Ytr, (Xte - mn) / st, Yte


@torch.no_grad()
def extract_all(bb, X, Y):
    ds  = TensorDataset(torch.from_numpy(X), torch.from_numpy(Y))
    ldr = DataLoader(ds, batch_size=512, shuffle=False,
                     pin_memory=USE_AMP, num_workers=2)
    fl, yl = [], []
    for bx, by in ldr:
        bx = bx.to(device, non_blocking=True)
        with amp_ctx():
            f = bb(bx)
        fl.append(f.float().cpu())
        yl.append(by)
    F_ = torch.cat(fl)
    Y_ = torch.cat(yl)
    return F_, Y_, Y_ // CLASSES_PER_TASK, Y_ % CLASSES_PER_TASK


# ==============================================================================
# MODULES
# ==============================================================================
class Expert(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(FEATURE_DIM, 256), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(256, 128),         nn.ReLU(), nn.Dropout(0.1),
            nn.Linear(128, CLASSES_PER_TASK)
        )
    def forward(self, x): return self.net(x)


class FlatRouter(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(FEATURE_DIM, 256), nn.ReLU(), nn.Dropout(0.1),
            nn.Linear(256, 128),         nn.ReLU(),
            nn.Linear(128, NUM_TASKS)
        )
    def forward(self, x): return self.net(x)


class HRouter(nn.Module):
    def __init__(self):
        super().__init__()
        self.root  = nn.Sequential(nn.Linear(FEATURE_DIM,128),nn.ReLU(),
                                   nn.Dropout(0.1),nn.Linear(128,2))
        self.sub_l = nn.Sequential(nn.Linear(FEATURE_DIM,128),nn.ReLU(),
                                   nn.Dropout(0.1),nn.Linear(128,5))
        self.sub_a = nn.Sequential(nn.Linear(FEATURE_DIM,128),nn.ReLU(),
                                   nn.Dropout(0.1),nn.Linear(128,5))
        self.register_buffer("liv", torch.tensor(LIVING_TASKS,   dtype=torch.long))
        self.register_buffer("art", torch.tensor(ARTIFACT_TASKS, dtype=torch.long))

    def forward(self, x, seen):
        pr = F.softmax(self.root(x),  -1)
        pl = F.softmax(self.sub_l(x), -1)
        pa = F.softmax(self.sub_a(x), -1)
        pt = torch.zeros(x.size(0), NUM_TASKS, device=x.device)
        pt[:, self.liv] = pr[:, 0:1] * pl
        pt[:, self.art] = pr[:, 1:2] * pa
        msk = torch.zeros(NUM_TASKS, device=x.device)
        msk[seen] = 1.0
        pt = pt * msk.unsqueeze(0)
        return pt / (pt.sum(-1, keepdim=True) + 1e-9)


# ==============================================================================
# PROTOTYPE BANK
# ==============================================================================
class ProtoBank:
    def __init__(self):
        self.cen={}; self.pmean={}; self.pV={}
        self.pmu={}; self.pivar={}; self.cache={}

    @torch.no_grad()
    def update(self, tid, f):
        f = f.to(device).float()
        self.cache[tid] = (torch.cat([self.cache[tid], f], 0)
                           if tid in self.cache else f.clone())
        A = self.cache[tid]
        n, d = A.shape
        self.cen[tid] = A.mean(0)
        k = min(MAHA_PCA_DIM, n-1, d)
        if k < 2: return
        try:
            mu = A.mean(0); Xc = A - mu
            _, _, V = torch.pca_lowrank(Xc, q=k, center=False)
            p = Xc @ V
            self.pmean[tid]=mu; self.pV[tid]=V
            self.pmu[tid]=p.mean(0)
            self.pivar[tid]=1.0/(p.var(0)+1e-4)
        except: pass

    @torch.no_grad()
    def cos(self, x, seen):
        S = torch.zeros(x.size(0), NUM_TASKS, device=x.device)
        v = [t for t in seen if t in self.cen]
        if v:
            P = torch.stack([self.cen[t] for t in v])
            S[:, v] = F.normalize(x,-1) @ F.normalize(P,-1).T
        return S

    @torch.no_grad()
    def maha(self, x, seen):
        S = torch.zeros(x.size(0), NUM_TASKS, device=x.device)
        for t in seen:
            if t not in self.pV: continue
            d = ((x - self.pmean[t]) @ self.pV[t]) - self.pmu[t]
            S[:, t] = -(d*d*self.pivar[t]).sum(1)
        return S


# ==============================================================================
# FUSION + CALIBRATION
# ==============================================================================
class Fusion:
    def __init__(self, wr=.45, wc=.20, wm=.15, we=.15, wen=.05):
        self.wr=wr; self.wc=wc; self.wm=wm; self.we=we; self.wen=wen

    @staticmethod
    def _zn(s, mf):
        mu = s.mean(0, keepdim=True)
        sd = s.std(0,  keepdim=True) + 1e-8
        return torch.sigmoid((s-mu)/sd) * mf.unsqueeze(0)

    def fuse(self, pr, sc, sm, se, sen, seen):
        mf = torch.zeros(NUM_TASKS, device=pr.device)
        mf[seen] = 1.0
        nr  = pr * mf.unsqueeze(0)
        nr  = nr / (nr.sum(-1, keepdim=True) + 1e-9)
        nc  = self._zn(sc,  mf)
        nm  = self._zn(sm,  mf)
        ne  = self._zn(se,  mf)
        nn_ = self._zn(sen, mf)
        f   = self.wr*nr + self.wc*nc + self.wm*nm + self.we*ne + self.wen*nn_
        f   = f * mf.unsqueeze(0)
        f   = f / (f.sum(-1, keepdim=True) + 1e-9)
        tops = torch.stack([nr.argmax(-1), nc.argmax(-1), nm.argmax(-1),
                            ne.argmax(-1), nn_.argmax(-1)], -1)
        ag = (tops == f.argmax(-1).unsqueeze(-1)).float().mean(-1)
        return f, ag


@torch.no_grad()
def calibrate(va_gpu, va_t, experts, router, rtype, seen,
              pbank, drop=None, power=CALIB_POWER):
    dev = va_gpu.device
    vt  = va_t.to(dev)
    m   = torch.isin(vt, torch.tensor(seen, device=dev))
    if int(m.sum()) < 100: return None
    Xv = va_gpu[m]; yv = vt[m]
    NEG = -1e9
    msk = torch.zeros(NUM_TASKS, device=dev)
    msk[seen] = 1.0
    if rtype == "flat":
        lg = router(Xv)
        mm = torch.full((NUM_TASKS,), NEG, device=dev)
        mm[seen] = 0
        pr = F.softmax(lg + mm.unsqueeze(0), -1)
    else:
        pr = router(Xv, seen)
    sc = pbank.cos( Xv, seen).masked_fill(msk==0, NEG)
    sm = pbank.maha(Xv, seen).masked_fill(msk==0, NEG)
    el   = torch.stack([experts[t](Xv) for t in seen], 0)
    sidx = torch.tensor(seen, device=dev)
    se   = torch.zeros(Xv.size(0), NUM_TASKS, device=dev)
    sn   = torch.zeros_like(se)
    se[:, sidx] = F.softmax(el,-1).max(-1).values.T
    sn[:, sidx] = torch.logsumexp(el,-1).T
    se = se.masked_fill(msk==0, NEG)
    sn = sn.masked_fill(msk==0, NEG)
    sigs = {"wr":pr,"wc":sc,"wm":sm,"we":se,"wen":sn}
    accs = {k:(s.argmax(-1)==yv).float().mean().item() for k,s in sigs.items()}
    if drop is not None: accs[drop] = 0.0
    w = {k: max(a,1e-3)**power for k,a in accs.items()}
    Z = sum(w.values())
    return {k:v/Z for k,v in w.items()}, accs


# ==============================================================================
# TOP-K
# ==============================================================================
def topk(fused, ag, seen):
    B = fused.size(0)
    K = min(3, len(seen))
    sp, st = torch.sort(fused, -1, descending=True)
    k = torch.ones(B, dtype=torch.long, device=fused.device)
    k[(sp[:,0] < TOP1_CONFIDENCE) | (ag < 0.6)] = min(2, K)
    k[(sp[:,0] < TOP2_CONFIDENCE) & (ag < 0.4)] = K
    return k, st[:, :K], sp


# ==============================================================================
# TRAINING
# ==============================================================================
def train_model(model, x, y, epochs, lr=3e-3, wd=1e-4,
                loss_fn=None, scaler=None):
    if loss_fn is None: loss_fn = nn.CrossEntropyLoss()
    if scaler  is None: scaler  = make_scaler()
    opt = optim.Adam(model.parameters(), lr=lr, weight_decay=wd)
    sch = optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    N   = x.size(0)
    for _ in range(epochs):
        model.train()
        perm = torch.randperm(N, device=device)
        for s in range(0, N, TRAIN_BATCH):
            bi = perm[s:s+TRAIN_BATCH]
            opt.zero_grad(set_to_none=True)
            with amp_ctx():
                loss = loss_fn(model(x[bi]), y[bi])
            scaler.scale(loss).backward()
            scaler.step(opt); scaler.update()
        sch.step()


def train_router(router, rtype, rx, ry, scaler):
    if rtype == "flat":
        train_model(router, rx, ry, ROUTER_EPOCHS, scaler=scaler)
        return
    art = torch.tensor(ARTIFACT_TASKS, device=device)
    yg  = torch.isin(ry, art).long()
    nl  = max((yg==0).sum().item(), 1)
    na  = max((yg==1).sum().item(), 1)
    wr  = torch.tensor([1.0/nl, 1.0/na], device=device)
    ml  = yg==0; ma = yg==1
    yll = torch.zeros(ml.sum(), dtype=torch.long, device=device)
    yal = torch.zeros(ma.sum(), dtype=torch.long, device=device)
    for i,t in enumerate(LIVING_TASKS):   yll[ry[ml]==t] = i
    for i,t in enumerate(ARTIFACT_TASKS): yal[ry[ma]==t] = i
    train_model(router.root,  rx,     yg,  ROUTER_EPOCHS,
                loss_fn=nn.CrossEntropyLoss(weight=wr), scaler=scaler)
    if ml.sum()>0: train_model(router.sub_l, rx[ml], yll, ROUTER_EPOCHS, scaler=scaler)
    if ma.sum()>0: train_model(router.sub_a, rx[ma], yal, ROUTER_EPOCHS, scaler=scaler)


# ==============================================================================
# INFERENCE — v4
# ==============================================================================
@torch.no_grad()
def infer_v4(te_gpu, yg, yt, experts, router, rtype, seen,
             use_verif, theta=0.15):
    router.eval()
    for e in experts.values(): e.eval()
    B   = te_gpu.size(0)
    t2r = {t:i for i,t in enumerate(seen)}
    preds = torch.empty(B, dtype=torch.long, device=device)
    for s0 in range(0, B, EVAL_BATCH):
        X   = te_gpu[s0:s0+EVAL_BATCH]; bsz = X.size(0)
        if rtype == "flat":
            lg = router(X)
            m  = torch.full((NUM_TASKS,), -1e9, device=device)
            m[seen] = 0
            pt = F.softmax(lg + m.unsqueeze(0), -1)
        else:
            pt = router(X, seen)
        sp, st = torch.sort(pt, -1, descending=True)
        t1 = st[:,0]
        t2 = st[:,1] if len(seen)>1 else st[:,0]
        mg = (sp[:,0]-sp[:,1] if len(seen)>1
              else torch.ones(bsz, device=device))
        el = torch.stack([experts[t](X) for t in seen], 0)
        ee = torch.logsumexp(el, -1)
        ch = t1.clone()
        if use_verif and len(seen)>1:
            for i in (mg<theta).nonzero(as_tuple=True)[0].tolist():
                c1=t1[i].item(); c2=t2[i].item()
                r1=t2r[c1];      r2=t2r[c2]
                p1=max(pt[i,c1].item(),1e-6)
                p2=max(pt[i,c2].item(),1e-6)
                if (np.log(p2)+0.5*ee[r2,i].item()) > (np.log(p1)+0.5*ee[r1,i].item()):
                    ch[i] = c2
        rows  = torch.tensor([t2r[t.item()] for t in ch], device=device)
        local = el[rows, torch.arange(bsz, device=device)].argmax(-1)
        preds[s0:s0+EVAL_BATCH] = ch*CLASSES_PER_TASK + local
    yg_ = yg.to(device); yt_ = yt.to(device)
    return {t: 100.0*(preds[yt_==t]==yg_[yt_==t]).float().mean().item()
            for t in seen if (yt_==t).any()}


# ==============================================================================
# INFERENCE — v5.1
# ==============================================================================
@torch.no_grad()
def infer_v5(te_gpu, yg, yt, experts, router, rtype, seen, pb, fusion):
    router.eval()
    for e in experts.values(): e.eval()
    B   = te_gpu.size(0)
    t2r = {t:i for i,t in enumerate(seen)}
    preds = torch.empty(B, dtype=torch.long, device=device)
    for s0 in range(0, B, EVAL_BATCH):
        X   = te_gpu[s0:s0+EVAL_BATCH]; bsz = X.size(0)
        if rtype == "flat":
            lg = router(X)
            m  = torch.full((NUM_TASKS,), -1e9, device=device)
            m[seen] = 0
            pr = F.softmax(lg + m.unsqueeze(0), -1)
        else:
            pr = router(X, seen)
        sc   = pb.cos( X, seen)
        sm   = pb.maha(X, seen)
        el   = torch.stack([experts[t](X) for t in seen], 0)
        sidx = torch.tensor(seen, device=device)
        se   = torch.zeros(bsz, NUM_TASKS, device=device)
        sn   = torch.zeros(bsz, NUM_TASKS, device=device)
        se[:, sidx] = F.softmax(el,-1).max(-1).values.T
        sn[:, sidx] = torch.logsumexp(el,-1).T
        fused, ag = fusion.fuse(pr, sc, sm, se, sn, seen)
        k, cands, sp = topk(fused, ag, seen)
        K  = cands.size(1)
        cf = torch.gather(fused, 1, cands)
        ce = torch.gather(sn,    1, cands)
        cs = (cf*ce).masked_fill(
            torch.arange(K,device=device).unsqueeze(0) >= k.unsqueeze(1),
            float("-inf"))
        bt    = torch.gather(cands,1,cs.argmax(1,keepdim=True)).squeeze(1)
        rows  = torch.tensor([t2r[t.item()] for t in bt], device=device)
        local = el[rows, torch.arange(bsz, device=device)].argmax(-1)
        preds[s0:s0+EVAL_BATCH] = bt*CLASSES_PER_TASK + local
    yg_ = yg.to(device); yt_ = yt.to(device)
    accs = {t: 100.0*(preds[yt_==t]==yg_[yt_==t]).float().mean().item()
            for t in seen if (yt_==t).any()}
    return accs, 100.0, 0


# ==============================================================================
# ONE CONTINUAL-LEARNING RUN
# ==============================================================================
def run_one(seed, order, M, variant, tr_f, tr_l, tr_t,
            te_gpu, te_g, te_t, scaler, va_gpu=None, va_t=None, wlog=None):
    set_seed(seed)
    if variant == "er":
        return _run_er(seed, order, M, tr_f, tr_l, tr_t, te_gpu, te_g, te_t, scaler)
    rtype     = "flat" if variant=="v4_flat" else "hierarchical"
    use_verif = variant in ("v4_hier","v5_hier")
    is_v5     = variant == "v5_hier"
    router    = (FlatRouter() if rtype=="flat" else HRouter()).to(device)
    experts   = {}
    pb        = ProtoBank()
    fusion    = Fusion()
    rx, ry    = [], []
    R         = np.zeros((NUM_TASKS, NUM_TASKS))
    for step, t in enumerate(order):
        msk = (tr_t == t)
        xt  = tr_f[msk].to(device)
        yl  = tr_l[msk].to(device)
        exp = Expert().to(device)
        train_model(exp, xt, yl, EXPERT_EPOCHS, scaler=scaler)
        exp.eval(); experts[t] = exp
        pb.update(t, xt)
        if M > 0:
            idx = torch.randperm(xt.size(0), device=device)[:M]
            rx.append(xt[idx])
            ry.append(torch.full((idx.numel(),), t, dtype=torch.long, device=device))
            crx = torch.cat(rx); cry = torch.cat(ry)
        else:
            crx = xt
            cry = torch.full((xt.size(0),), t, dtype=torch.long, device=device)
        train_router(router, rtype, crx, cry, scaler)
        seen = list(order[:step+1])
        if is_v5 and va_gpu is not None:
            out = calibrate(va_gpu, va_t, experts, router, rtype, seen, pb)
            if out is not None:
                w, acc = out
                fusion.wr=w["wr"]; fusion.wc=w["wc"]; fusion.wm=w["wm"]
                fusion.we=w["we"]; fusion.wen=w["wen"]
                if wlog is not None:
                    wlog.append(dict(M=M, seed=seed, step=step,
                        wr=w["wr"],wc=w["wc"],wm=w["wm"],we=w["we"],wen=w["wen"],
                        router_acc=acc["wr"],cosine_acc=acc["wc"],
                        maha_acc=acc["wm"],expert_acc=acc["we"],energy_acc=acc["wen"]))
        if is_v5:
            ad, _, _ = infer_v5(te_gpu, te_g, te_t, experts, router, rtype, seen, pb, fusion)
        else:
            ad = infer_v4(te_gpu, te_g, te_t, experts, router, rtype, seen, use_verif)
        for s, a in ad.items(): R[step, s] = a
    fa  = float(np.mean([R[NUM_TASKS-1, s] for s in order]))
    bwt = float(np.mean([R[NUM_TASKS-1, order[j]] - R[j, order[j]]
                          for j in range(NUM_TASKS-1)]))
    return fa, bwt, 100.0, 0


# ==============================================================================
# EXPERIENCE REPLAY BASELINE
# ==============================================================================
def _run_er(seed, order, M, tr_f, tr_l, tr_t, te_gpu, te_g, te_t, scaler):
    set_seed(seed)
    model = nn.Sequential(
        nn.Linear(FEATURE_DIM,256), nn.ReLU(), nn.Dropout(0.1),
        nn.Linear(256, 100)
    ).to(device)
    R = np.zeros((NUM_TASKS, NUM_TASKS))
    bx, by = [], []
    for step, t in enumerate(order):
        msk = (tr_t == t)
        xt  = tr_f[msk].to(device)
        yl  = (tr_l[msk] + t*CLASSES_PER_TASK).to(device)
        if M > 0:
            idx = torch.randperm(xt.size(0), device=device)[:M]
            bx.append(xt[idx]); by.append(yl[idx])
            tx = torch.cat(bx);  ty = torch.cat(by)
        else:
            tx = xt; ty = yl
        train_model(model, tx, ty, EXPERT_EPOCHS, scaler=scaler)
        model.eval()
        with torch.no_grad():
            preds = model(te_gpu).argmax(-1)
        yg = te_g.to(device); yt = te_t.to(device)
        for s in order[:step+1]:
            m = (yt==s)
            if m.any():
                R[step,s] = 100.0*(preds[m]==yg[m]).float().mean().item()
    fa  = float(np.mean([R[NUM_TASKS-1,s] for s in order]))
    bwt = float(np.mean([R[NUM_TASKS-1,order[j]]-R[j,order[j]]
                          for j in range(NUM_TASKS-1)]))
    return fa, bwt, 100.0, 0


# ==============================================================================
# CALIBRATED ABLATION
# ==============================================================================
CALIB_CFG = {
    "Full AEFR":None,"- Router":"wr","- Cosine":"wc",
    "- Mahalanobis":"wm","- Expert conf":"we","- Energy":"wen"
}
MANUAL_CFG = {
    "Router only":  dict(wr=1.0,wc=0.0,wm=0.0,we=0.0,wen=0.0),
    "Cosine only":  dict(wr=0.0,wc=1.0,wm=0.0,we=0.0,wen=0.0),
    "Mahal only":   dict(wr=0.0,wc=0.0,wm=1.0,we=0.0,wen=0.0),
    "Expert only":  dict(wr=0.0,wc=0.0,wm=0.0,we=1.0,wen=0.0),
    "Energy only":  dict(wr=0.0,wc=0.0,wm=0.0,we=0.0,wen=1.0),
    "Prototype only":dict(wr=0.0,wc=0.5,wm=0.5,we=0.0,wen=0.0),
    "Proto+Expert": dict(wr=0.0,wc=0.3,wm=0.3,we=0.4,wen=0.0),
}


def abl_run(seed, tr_f, tr_l, tr_t, te_gpu, te_g, te_t,
            va_gpu, va_t, fw, drop, scaler):
    set_seed(seed)
    order  = TASK_ORDERS["Order_A"]
    router = HRouter().to(device)
    experts= {}; pb = ProtoBank()
    fusion = Fusion(**fw) if fw else Fusion()
    R = np.zeros((NUM_TASKS, NUM_TASKS))
    for step, t in enumerate(order):
        msk = (tr_t==t)
        xt  = tr_f[msk].to(device); yl = tr_l[msk].to(device)
        exp = Expert().to(device)
        train_model(exp, xt, yl, EXPERT_EPOCHS, scaler=scaler)
        exp.eval(); experts[t] = exp
        pb.update(t, xt)
        cry = torch.full((xt.size(0),), t, dtype=torch.long, device=device)
        train_router(router, "hierarchical", xt, cry, scaler)
        seen = list(order[:step+1])
        if fw is None:
            out = calibrate(va_gpu, va_t, experts, router,
                            "hierarchical", seen, pb, drop=drop)
            if out is not None:
                w, _ = out
                fusion.wr=w["wr"]; fusion.wc=w["wc"]; fusion.wm=w["wm"]
                fusion.we=w["we"]; fusion.wen=w["wen"]
        ad, _, _ = infer_v5(te_gpu, te_g, te_t, experts, router,
                            "hierarchical", seen, pb, fusion)
        for s, a in ad.items(): R[step,s] = a
    fa = float(np.mean([R[NUM_TASKS-1,s] for s in order]))
    pt = {s: R[NUM_TASKS-1,s] for s in order}
    return fa, pt


def run_ablation(tr_f, tr_l, tr_t, te_gpu, te_g, te_t, va_gpu, va_t):
    log("\n"+"="*80)
    log(" CALIBRATED ABLATION (M=0, Order_A, 3 seeds)")
    log("="*80)
    seeds  = [42, 100, 2024]
    scaler = make_scaler()
    res    = {}
    allc   = (list(CALIB_CFG.items()) +
              [(n,(w,None)) for n,w in MANUAL_CFG.items()])
    done   = 0; total = len(allc)*len(seeds); wall = time.time()
    for name, cfg in allc:
        fw, drop = (None, cfg) if name in CALIB_CFG else (cfg, None)
        accs=[]; pts=[]
        for s in seeds:
            fa, pt = abl_run(s, tr_f, tr_l, tr_t, te_gpu, te_g, te_t,
                             va_gpu, va_t, fw, drop, scaler)
            accs.append(fa); pts.append(pt); done+=1
        res[name] = {
            "mean": float(np.mean(accs)),
            "std":  float(np.std(accs)),
            "per_task": {str(t): float(np.mean([p[t] for p in pts]))
                         for t in TASK_ORDERS["Order_A"]}
        }
        eta = (time.time()-wall)/done*(total-done)
        log(f"  {name:<16} {res[name]['mean']:5.2f}±{res[name]['std']:.2f}% "
            f"[ETA {eta/60:.1f}min | {done}/{total}]")
    base = res["Full AEFR"]["mean"]
    rout = res["Router only"]["mean"]
    log(f"\n  {'Config':<16} {'Acc':>8} {'ΔFull':>8}")
    log(f"  {'-'*16} {'-'*8} {'-'*8}")
    for n,_ in allc:
        r=res[n]; log(f"  {n:<16} {r['mean']:>6.2f}% {r['mean']-base:>+7.2f}%")
    log(f"\n  Full(calibrated): {base:.2f}% | Router only: {rout:.2f}% | "
        f"gain: {base-rout:+.2f}%")
    log("="*80)
    return res


# ==============================================================================
# FIGURES
# ==============================================================================
def make_figures(all_res, wlog):
    import matplotlib; matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    VN={"er":"ER","v4_flat":"v4-Flat","v4_hier":"v4-Hier","v5_hier":"v5.1-CAL"}
    MK={"er":"o","v4_flat":"s","v4_hier":"^","v5_hier":"D"}
    CL={"er":"gray","v4_flat":"#1f77b4","v4_hier":"#2ca02c","v5_hier":"#d62728"}
    fig,axes=plt.subplots(1,2,figsize=(13,4.5))
    for ax,on in zip(axes,["Order_A","Order_B"]):
        for v in VN:
            mus=[np.mean(all_res[M][on][v]["accs"]) for M in MEMORY_LEVELS]
            sds=[np.std( all_res[M][on][v]["accs"]) for M in MEMORY_LEVELS]
            ax.errorbar(MEMORY_LEVELS,mus,yerr=sds,marker=MK[v],color=CL[v],
                        label=VN[v],capsize=3,linewidth=2,markersize=6)
        ax.set_xlabel("Memory M (samples/task)"); ax.set_ylabel("Final Avg Acc (%)")
        ax.set_title(on); ax.legend(fontsize=8); ax.grid(alpha=0.3)
    plt.suptitle("Split CIFAR-100 | Frozen ResNet-18 | Accuracy vs Memory")
    plt.tight_layout()
    p1=os.path.join(OUTPUT_DIR,"fig1_accuracy_vs_memory.png")
    plt.savefig(p1,dpi=150,bbox_inches="tight"); plt.close()
    log(f"📊 Figure 1 → {p1}")
    if wlog:
        fig,axes=plt.subplots(1,2,figsize=(13,4.5))
        for M in MEMORY_LEVELS:
            ws=[w for w in wlog if w["M"]==M]
            if not ws: continue
            by_step={}
            for w in ws: by_step.setdefault(w["step"],[]).append(w["wr"])
            steps=sorted(by_step)
            mus=[np.mean(by_step[s]) for s in steps]
            sds=[np.std( by_step[s]) for s in steps]
            axes[0].errorbar(steps,mus,yerr=sds,marker="o",capsize=2,
                             label=f"M={M}",linewidth=2,markersize=4)
            by_step2={}
            for w in ws: by_step2.setdefault(w["step"],[]).append(w["router_acc"])
            mus2=[np.mean(by_step2[s]) for s in sorted(by_step2)]
            axes[1].plot(sorted(by_step2),mus2,marker="s",
                         label=f"M={M}",linewidth=2,markersize=4)
        axes[0].set_xlabel("CL Step"); axes[0].set_ylabel("Router weight wr")
        axes[0].set_title("Router trust vs experience")
        axes[0].legend(); axes[0].grid(alpha=0.3)
        axes[1].set_xlabel("CL Step"); axes[1].set_ylabel("Router val accuracy")
        axes[1].set_title("Router quality drives its weight")
        axes[1].legend(); axes[1].grid(alpha=0.3)
        plt.tight_layout()
        p2=os.path.join(OUTPUT_DIR,"fig2_router_weight_trajectory.png")
        plt.savefig(p2,dpi=150,bbox_inches="tight"); plt.close()
        log(f"📊 Figure 2 → {p2}")


# ==============================================================================
# MAIN
# ==============================================================================
def main():
    log("="*90)
    log(f" CGL-AARM v5.1 CALIBRATED — FULL RUN | {device}")
    log(f" M={MEMORY_LEVELS} | {len(SEEDS)} seeds × 2 orders | "
        f"Ablation={'ON' if RUN_ABLATION else 'OFF'}")
    log("="*90)

    Xtr, Ytr, Xte, Yte = load_cifar100()
    log(f"  Data: train={Xtr.shape} test={Xte.shape}")

    rng=np.random.default_rng(0); tri=[]; vai=[]
    for c in range(100):
        ci=np.where(Ytr==c)[0]; rng.shuffle(ci)
        sp=int(VAL_FRAC*len(ci)); vai.extend(ci[:sp]); tri.extend(ci[sp:])

    log("Extracting features...")
    bb = torchvision.models.resnet18(
        weights=torchvision.models.ResNet18_Weights.DEFAULT)
    bb.fc = nn.Identity(); bb = bb.to(device)
    for p in bb.parameters(): p.requires_grad_(False)
    bb.eval()

    tr_f,_,tr_t,tr_l = extract_all(bb, Xtr[tri], Ytr[tri])
    va_f,va_g,va_t,_ = extract_all(bb, Xtr[vai], Ytr[vai])
    te_f,te_g,te_t,_ = extract_all(bb, Xte, Yte)
    del bb
    if USE_AMP: torch.cuda.empty_cache()

    te_gpu = te_f.to(device)
    va_gpu = va_f.to(device)
    log(f"  tr={tr_f.shape} va={va_f.shape} te={te_f.shape}")

    scaler   = make_scaler()
    VARIANTS = ["er","v4_flat","v4_hier","v5_hier"]
    VN       = {"er":"ER","v4_flat":"v4-Flat",
                "v4_hier":"v4-Hier","v5_hier":"v5.1-CAL🔥"}
    all_res  = {}; wlog = []; wall = time.time()

    for M in MEMORY_LEVELS:
        all_res[M] = {}
        log(f"\n{'='*90}\n M={M}\n{'='*90}")
        for on, oseq in TASK_ORDERS.items():
            all_res[M][on] = {v:{"accs":[],"bwts":[]} for v in VARIANTS}
            for si, s in enumerate(SEEDS):
                t0  = time.time(); row = []
                for v in VARIANTS:
                    a,b,_,_ = run_one(s, oseq, M, v, tr_f, tr_l, tr_t,
                                      te_gpu, te_g, te_t, scaler,
                                      va_gpu, va_t,
                                      wlog if v=="v5_hier" else None)
                    all_res[M][on][v]["accs"].append(a)
                    all_res[M][on][v]["bwts"].append(b)
                    row.append(f"{VN[v]}:{a:.1f}%")
                log(f"  M={M:3d} {on} s{s} ({si+1}/{len(SEEDS)}) | "
                    + " | ".join(row) + f" | {time.time()-t0:.0f}s")

    log(f"\n{'='*95}\n FINAL RESULTS\n{'='*95}")
    for M in MEMORY_LEVELS:
        log(f"\n M={M} samples/task")
        log(f"  {'Method':<12} {'Order_A':>18} {'Order_B':>18} {'BWT':>10}")
        log(f"  {'-'*12} {'-'*18} {'-'*18} {'-'*10}")
        for v in VARIANTS:
            ra=all_res[M]["Order_A"][v]; rb=all_res[M]["Order_B"][v]
            sa=f"{np.mean(ra['accs']):.2f}±{np.std(ra['accs']):.2f}%"
            sb=f"{np.mean(rb['accs']):.2f}±{np.std(rb['accs']):.2f}%"
            mb=(np.mean(ra["bwts"])+np.mean(rb["bwts"]))/2
            log(f"  {VN[v]:<12} {sa:>18} {sb:>18} {mb:>+9.2f}%")
        for on in TASK_ORDERS:
            h4=all_res[M][on]["v4_hier"]["accs"]
            h5=all_res[M][on]["v5_hier"]["accs"]
            try: _,p=stats.ttest_rel(h5,h4)
            except: p=1.0
            log(f"   v5 vs v4 [{on}] Δ={np.mean(h5)-np.mean(h4):+.2f}% "
                f"p={p:.4f} {'✅' if p<0.05 else '○'}")

    log(f"\n{'='*95}\n SUMMARY\n{'='*95}")
    for M in MEMORY_LEVELS:
        er=np.mean([np.mean(all_res[M][on]["er"]["accs"])       for on in TASK_ORDERS])
        v4=np.mean([np.mean(all_res[M][on]["v4_hier"]["accs"])  for on in TASK_ORDERS])
        v5=np.mean([np.mean(all_res[M][on]["v5_hier"]["accs"])  for on in TASK_ORDERS])
        b5=np.mean([np.mean(all_res[M][on]["v5_hier"]["bwts"])  for on in TASK_ORDERS])
        log(f"  M={M:3d}: ER={er:.2f}% v4={v4:.2f}% v5.1={v5:.2f}% "
            f"BWT={b5:+.2f}% | Δ(v5-ER)={v5-er:+.2f}% Δ(v5-v4)={v5-v4:+.2f}%")

    log("\n Router calibrated weight at FINAL step (by M):")
    for M in MEMORY_LEVELS:
        ws=[w["wr"]         for w in wlog if w["M"]==M and w["step"]==NUM_TASKS-1]
        ra=[w["router_acc"] for w in wlog if w["M"]==M and w["step"]==NUM_TASKS-1]
        if ws: log(f"  M={M:3d}: wr={np.mean(ws):.3f}±{np.std(ws):.3f} "
                   f"(router val-acc={np.mean(ra):.2f})")

    abl = run_ablation(tr_f,tr_l,tr_t,te_gpu,te_g,te_t,va_gpu,va_t) \
          if RUN_ABLATION else None

    try: make_figures(all_res, wlog)
    except Exception as e: log(f"[figure warn] {e}")

    out = {
        "benchmark": {str(M): {on: {v: all_res[M][on][v] for v in VARIANTS}
                                for on in TASK_ORDERS}
                      for M in MEMORY_LEVELS},
        "weight_log": wlog,
        "ablation":   abl
    }
    p = os.path.join(OUTPUT_DIR, "v51_results.json")
    with open(p,"w") as f: json.dump(out, f, indent=2, default=float)
    log(f"\n💾 Saved → {p}")

    if os.path.isdir("/content/drive/MyDrive"):
        import shutil
        dd="/content/drive/MyDrive/CGL_AARM_v5"
        os.makedirs(dd, exist_ok=True)
        shutil.copy(p, os.path.join(dd,"v51_results.json"))
        for fig in ["fig1_accuracy_vs_memory.png","fig2_router_weight_trajectory.png"]:
            src=os.path.join(OUTPUT_DIR,fig)
            if os.path.isfile(src): shutil.copy(src,os.path.join(dd,fig))
        log(f"💾 Also saved → {dd}/")

    log(f"\n Total time: {(time.time()-wall)/60:.1f} min")
    log("✅ DONE — paste output for interpretation")


# ==============================================================================
# RUN
# ==============================================================================
main()
```
